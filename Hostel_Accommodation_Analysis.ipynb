### 1. Ingest Dataset & Import Libraries

```python
import os
import sqlite3
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset from CSV or SQLite
df = pd.read_csv('hostel_dataset.csv')
df.head()
```

### 2. Data Cleaning & Feature Engineering (Unit 4 Syllabus)

```python
# Impute missing values and compute derived ratios
df['available_beds'] = np.maximum(0, df['capacity'].fillna(0) - df['occupancy'].fillna(0))
df['occupancy_rate_pct'] = np.where(df['capacity'] > 0, (df['occupancy'] / df['capacity']) * 100.0, 0.0)
df['fee_clearance_ratio'] = np.where(df['total_fee'] > 0, df['paid_amount'] / df['total_fee'], 0.0)
df.info()
```

### 3. Descriptive Statistics & Outlier Detection (Unit 1 & 3 Syllabus)

```python
num_cols = ['year_of_study', 'capacity', 'occupancy', 'total_fee', 'paid_amount', 'pending_amount']
stats_summary = df[num_cols].describe().T
stats_summary['IQR'] = df[num_cols].quantile(0.75) - df[num_cols].quantile(0.25)
stats_summary
```

### 4. Data Visualizations (Unit 5 Syllabus)

```python
# Visual 1: Bed Capacity vs Occupancy by Room Type
room_agg = df[df['room_type'] != 'Unassigned'].groupby('room_type')[['capacity', 'occupancy']].sum()
room_agg.plot(kind='bar', figsize=(8, 5), color=['#2b5c8f', '#e76f51'], edgecolor='black')
plt.title('Hostel Bed Capacity vs. Current Occupancy by Room Type')
plt.ylabel('Beds')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Visual 2: Fee Distribution Across Years of Study (Box Plot)
plt.figure(figsize=(8, 5))
sns.boxplot(x='year_of_study', y='paid_amount', data=df, palette='Set2')
plt.title('Dispersion of Student Fee Payments Across Years of Study')
plt.xlabel('Year of Study')
plt.ylabel('Fee Paid (INR)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Visual 3: Correlation Matrix Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Matrix of Numerical Hostel Attributes')
plt.show()
```
